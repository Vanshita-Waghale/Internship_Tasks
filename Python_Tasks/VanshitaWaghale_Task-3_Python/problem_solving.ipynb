## Q1. Find the Frequency of Each Word in a Sentence
Pseudocode:
1. Take a sentence as input.
2. Split the sentence into individual words.
3. Check each word in the sentence.
4. If the word already exists, increase its count by 1.
5. Otherwise, add the word with a count of 1.
6. Return the dictionary containing each word and its frequency.

```python
def word_frequency(sentence):
    word_count = {}
    words = sentence.split()

    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    return word_count

# Test Case 1
print(word_frequency("python is easy and python is powerful."))

# Test Case 2
print(word_frequency("practices makes perfect and perfect is the key to success"))
```

### Q2. Find the Frequency of Each Character in a String
Pseudocode:
1. Take a string as input.
2. Create an empty dictionary to store character counts.
3. Check each character in the string.
4. If the character already exists, increase its count by 1.
5. Otherwise, add the character with a count of 1.
6. Return the dictionary containing each character and its frequency.

```python
def char_frequency(string):
    char_count = {}

    for char in string:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

    return char_count

# Test Case 1
print(char_frequency("banana"))

# Test Case 2
print(char_frequency("frequency"))
```

### Q3. Find the First Non-Repeating Character in a Word
Pseudocode:
1. Take a word as input.
2. Count how many times each character appears.
3. Check the characters again in their original order.
4. Find the first character whose count is 1.
5. Return that character.
6. If no such character exists, return None.

```python
def first_non_repeating_char(word):
    char_count = {}

    for char in word:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

    for char in word:
        if char_count[char] == 1:
            return char

    return None

# Test Case 1
print(first_non_repeating_char("banana"))

# Test Case 2
print(first_non_repeating_char("Aarti"))
```

### Q4. Check Whether Two Strings Are Anagrams
Pseudocode:
1. Take two strings as input.
2. Convert both strings to lowercase.
3. Sort the characters of both strings.
4. Compare the sorted strings.
5. If they are equal, return True.
6. Otherwise, return False.

```python
def are_anagrams(str1, str2):
    return sorted(str1.lower()) == sorted(str2.lower())

# Test Case 1
print(are_anagrams("listen", "silent"))

# Test Case 2
print(are_anagrams("hello", "world"))
```

### Q5. Find the First Repeating Character in a String
Pseudocode:
1. Take a string as input.
2. Count how many times each character appears.
3. Check the characters again in their original order.
4. Find the first character whose count is greater than 1.
5. Return that character.
6. If no character repeats, return None.

```python
def first_repeating_char(string):
    char_count = {}

    for char in string:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

    for char in string:
        if char_count[char] > 1:
            return char

    return None

# Test Case 1
print(first_repeating_char("programming"))

# Test Case 2
print(first_repeating_char("banana"))
```

### Q6. Move All Zeros to the End of a List
Pseudocode:
1. Take a list of numbers as input.
2. Collect all non-zero numbers in a new list.
3. Count the number of zeros in the original list.
4. Add the zeros to the end of the new list.
5. Return the updated list.

```python
def move_zeros_to_end(lst):
    non_zero_elements = [x for x in lst if x != 0]
    zero_count = lst.count(0)

    return non_zero_elements + [0] * zero_count

# Test Case 1
print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))

# Test Case 2
print(move_zeros_to_end([1, 0, 1, 2, 0, 1, 34, 0]))
```

### Q7. Find the Second Highest Distinct Number in a List
Pseudocode:
1. Take a list of numbers as input.
2. Remove duplicate numbers.
3. If fewer than two distinct numbers exist, return None.
4. Sort the numbers in descending order.
5. Return the second number in the sorted list.

```python
def second_highest(lst):
    distinct_numbers = list(set(lst))

    if len(distinct_numbers) < 2:
        return None

    distinct_numbers.sort(reverse=True)

    return distinct_numbers[1]

# Test Case 1
print(second_highest([10, 20, 40, 30, 40, 20]))

# Test Case 2
print(second_highest([5, 5, 5, 5]))
```

### Q8. Remove Duplicates Without Changing the Original Order
Pseudocode:
1. Take a list as input.
2. Create an empty list to store unique elements.
3. Check each element in the original list.
4. If the element has not appeared before, add it to the new list.
5. Skip elements that have already appeared.
6. Return the list without duplicates.

```python
def remove_duplicates(lst):
    seen = set()
    result = []

    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)

    return result

# Test Case 1
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Test Case 2
print(remove_duplicates([1, 2, 3, 2, 1, 4]))
```

### Q9. Find the Missing Number from 1 to N
Pseudocode:
1. Take a list of numbers and the value of N as input.
2. Calculate the expected sum of numbers from 1 to N.
3. Calculate the sum of the numbers in the given list.
4. Subtract the actual sum from the expected sum.
5. Return the result as the missing number.

```python
def find_missing_number(lst, N):
    expected_sum = N * (N + 1) // 2
    actual_sum = sum(lst)

    return expected_sum - actual_sum

# Test Case 1
print(find_missing_number([1, 2, 4, 5, 6], 6))

# Test Case 2
print(find_missing_number([1, 2, 4, 6, 3, 7, 8], 8))
```

### Q10. Find Two Numbers Whose Sum Equals a Target
Pseudocode:
1. Take a list of numbers and a target value as input.
2. Create an empty set to store previously checked numbers.
3. For each number, calculate the number needed to reach the target.
4. Check whether that number exists in the set.
5. If it exists, return both numbers as a pair.
6. Otherwise, add the current number to the set.
7. If no pair is found, return None.

```python
def find_two_numbers_with_sum(lst, target):
    num_set = set()

    for num in lst:
        complement = target - num

        if complement in num_set:
            return (complement, num)

        num_set.add(num)

    return None

# Test Case 1
print(find_two_numbers_with_sum([2, 7, 11, 15], 9))

# Test Case 2
print(find_two_numbers_with_sum([1, 2, 3, 4, 5, 6, 7, 5], 8))
```

### Q11. Find the Number Closest to Zero
Pseudocode:
1. Take a list of numbers as input.
2. Start with no closest number.
3. Check each number in the list.
4. Compare its absolute value with the absolute value of the current closest number.
5. If it is smaller, update the closest number.
6. Return the number closest to zero.

```python
def find_closest_to_zero(lst):
    closest = None

    for num in lst:
        if closest is None or abs(num) < abs(closest):
            closest = num

    return closest

# Test Case 1
print(find_closest_to_zero([-8, -3, 2, 5, -2]))

# Test Case 2
print(find_closest_to_zero([1, 2, 3, 4, 5]))
```

### Q12. Find Common Elements Between Two Lists
Pseudocode:
1. Take two lists as input.
2. Convert both lists into sets.
3. Find the elements that exist in both sets.
4. Convert the common elements into a list.
5. Return the list of common elements.

```python
def find_common_elements(lst1, lst2):
    return list(set(lst1) & set(lst2))

# Test Case 1
print(find_common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Test Case 2
print(find_common_elements([10, 20, 30], [40, 50, 60]))
```

### Q13. Rotate a List to the Right by K Positions
Pseudocode:
1. Take a list and the number of rotations K as input.
2. Calculate the effective rotations using K modulo the list length.
3. Separate the last K elements from the list.
4. Move these elements to the beginning.
5. Add the remaining elements after them.
6. Return the rotated list.

```python
def rotate_list(lst, k):
    k = k % len(lst)

    return lst[-k:] + lst[:-k]

# Test Case 1
print(rotate_list([1, 2, 3, 4, 5], 2))

# Test Case 2
print(rotate_list([10, 20, 30, 40, 50], 3))
```

### Q14. Find the Maximum Sum of Two Consecutive Elements
Pseudocode:
1. Take a list of numbers as input.
2. If the list has fewer than two elements, return None.
3. Calculate the sum of the first two elements.
4. Check each pair of consecutive elements.
5. Calculate the sum of each pair and compare it with the maximum sum.
6. Update the maximum sum whenever a larger sum is found.
7. Return the maximum sum.

```python
def max_sum_of_consecutive_elements(lst):
    if len(lst) < 2:
        return None

    max_sum = lst[0] + lst[1]

    for i in range(1, len(lst) - 1):
        current_sum = lst[i] + lst[i + 1]

        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum

# Test Case 1
print(max_sum_of_consecutive_elements([4, 7, 2, 9, 5]))

# Test Case 2
print(max_sum_of_consecutive_elements([1, 2, 3, 4, 5]))
```

### Q15. Find All Pairs of Numbers Whose Sum Equals a Target
Pseudocode:
1. Take a list of numbers and a target value as input.
2. Create an empty list to store pairs and a set to store previously checked numbers.
3. For each number, calculate the number needed to reach the target.
4. If that number exists in the set, add the pair to the result list.
5. Add the current number to the set.
6. Continue until all numbers have been checked.
7. Return the list of pairs found.

```python
def find_all_pairs_with_sum(lst, target):
    pairs = []
    seen = set()

    for num in lst:
        complement = target - num

        if complement in seen:
            pairs.append((complement, num))

        seen.add(num)

    return pairs

# Test Case 1
print(find_all_pairs_with_sum([2, 4, 3, 5, 7, 8, 1], 9))

# Test Case 2
print(find_all_pairs_with_sum([1, 2, 3, 4, 5], 5))
```
