```python
# Q35
import copy
a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
#output = [[1, 99], [3, 4]] because copy.copy() creates a shallow copy of the list, so changes to the inner lists in b affect a.
```

```python
# Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
#output = [[1, 2], [3, 4]] because deepcopy creates a new copy of the inner lists, so changes to b do not affect a.
```

```python
# Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
#output = [[10], [20, 30]] because a.copy() creates a shallow copy of the list, so changes to the inner lists in a affect b.
```

```python
# Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
#output = [[10], [20]] because deepcopy creates a new copy of the inner lists, so changes to a do not affect b.
```

```python
# Q39
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
#output = False and True because copy.copy() creates a shallow copy of the list, so a and b are different objects, but their elements refer to the same inner lists.
```

```python
# Q44
def test(*args):
    print(args)
test(1, 2)
#output = (1, 2) because args is a tuple that collects positional arguments passed to the function.
```

```python
# Q45
def test(*args):
    print(type(args))

test("a", "b", "c")
#output = <class 'tuple'> because args is a tuple that collects positional arguments passed to the function.
```

```python
# Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
#output = {'name': 'Ram', 'age': 25} because kwargs is a dictionary that collects keyword arguments passed to the function.
```
