If its frequency is one and its length is greater than the current longest word's length 
-> Update the longest word.

If the length is equal -> Keep the first word.

End of loop -> Display the longest word.

5) Code Syntax:

sentence = "Python makes coding useful Python"
words = sentence.lower().split()
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

longest = ""

for word in words:
    if frequency[word] == 1 and len(word) > len(longest):
        longest = word

print(longest)
